```python
def send_mail():
    # 1. DEFINE VARIABLES (Easily configurable for the user)
    my_email = 'YOUR_EMAIL@gmail.com'        # Replace with your email
    my_password = 'YOUR_APP_PASSWORD'        # Replace with Google "App Password" (not your normal password)
    receiver_email = 'RECEIVER_EMAIL@gmail.com' # The email receiving the alert

    # 2. CONNECTION
    server = smtplib.SMTP_SSL('smtp.gmail.com', 465)
    server.ehlo()
    # server.starttls() # Not needed with SMTP_SSL
    server.ehlo()

    # 3. LOGIN (Using the variables)
    server.login(my_email, my_password)

    # 4. PREPARE MESSAGE
    subject = "The Shirt you want is below $15! Now is your chance to buy!"
    body = "Zayd, This is the moment! You can buy it at a lower price."

    # Standard format for emails (Subject + \n\n + Body)
    msg = f"Subject: {subject}\n\n{body}"

    # 5. SEND EMAIL
    server.sendmail(
        my_email,       # Sender
        receiver_email, # Receiver
        msg
    )

    # It is good practice to close the connection
    server.quit()
```
